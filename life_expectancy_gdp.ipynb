```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
all_data = pd.read_csv('all_data.csv')
all_data.head()
```

```python
all_data.info()
```

```python
#Has life expectancy increased over time in these 6 nations (2 quantitative variables and 1 categorical variable - scatterplot | heatmap)
sns.lineplot(x = all_data['Year'],
                y = all_data['Life expectancy at birth (years)'],
                hue=all_data['Country'],
                palette='bright')
plt.title('Life expectancy over the years in 6 Nations')
plt.xlabel('Year')
plt.ylabel('Life Expectancy (years)')
plt.xticks(rotation=45)
plt.legend(title='Country')
plt.legend()
plt.show()
```

```python
#Looking at this lineplot above, we can see that all of these nations have had in increase in life expectancy at 
#birth over the years from 2000 to 2016. Zimbabwe started with a very low life expectency but arount 2007 have had
#an accelerating increase to a life expectency at 60 whilst the other nations have had a steady growth.
```

```python
#Is there a correlation between GDP and life expectency over these nations
sns.scatterplot(x = all_data['GDP'],
                y = all_data['Life expectancy at birth (years)'],
                hue = all_data['Country'])
plt.title('GDP vs Life expectancy')
plt.xlabel('GDP')
plt.ylabel('Life Expectancy (years)')
plt.xticks(rotation=45)
plt.legend(title='Country')
plt.figure(figsize=(12,6))
plt.show()
```

```python
#Average life expectancy of these nations (catergorical variable vs quantitative variable)
sns.barplot(x = 'Country', y = 'Life expectancy at birth (years)', data = all_data)
plt.title('Life expectancy for each Nation')
plt.xlabel('Country')
plt.xticks(rotation=45)
plt.figure(figsize=(6,12))
plt.show()
```

```python
#The barchart above shows us that Germany has the highest average life expectancy at birth but the average is pretty
#close to Chile, China, Mexico and the United States at just below 80 years old. Zimbabwe has the lowest average and
#a significant drop from the other nations at an average of just 50 years of life expectancy.
```

```python
#What is the disrtibution of that life expectancy
sns.histplot(all_data['Life expectancy at birth (years)'])
plt.title('Distribution of life expectancy (years)')
plt.xlabel('Life expectancy (years)')
plt.ylabel('Frequency')
```

```python
#Looking at the distribution of this the life expectancy at birth in years, we can see that it is negatively skewed
#rather than normally distributed. This is due to 5 of the 6 nations having a high average life expectancy whereas 
#Zimbabwe has a comparetively low average life expectancy, lowering the average. This being an anomaly hugely affects
#the distribution of the graph.
```
